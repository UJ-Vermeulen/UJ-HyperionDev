[Cars93 metadata](https://jse.amstat.org/v1n1/datasets.lock.html) 

```python
# Import libraries
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Load data
df = pd.read_csv('Cars93.csv', index_col = 0, keep_default_na=False)
```

### 1. Boxplot

```python

# Boxplot for revs per mile
# Rotate the x-label to make it more readable and prevent it from 
# being print over each other. 
df.boxplot(column="Rev.per.mile", by="Manufacturer", rot=90)

# Enter the details of the graph
plt.title("Boxplot of Rev.per.mile by Manufacturer")
plt.suptitle("")
plt.xlabel("Manufacturer")
plt.ylabel("Rev.per.mile")
plt.show()
```

**Boxplot question:**  
Which of these manufacturers has the car with the highest revs per mile?

Answer: The manufacturer with the highest revs per mile is 'Geo', as seen in the boxplot figure above.

### 2. Histogram

```python
# Histogram of MPG
plt.hist(df["MPG.city"], label="MPG.city")
plt.hist(df['MPG.highway'], label="MPG.highway")

# Enter the details of the graph
plt.title("Histogram of City vs. Highway MPG")
plt.suptitle("")
plt.xlabel("Miles Per Gallon (MPG)")
plt.ylabel("Frequency")
plt.legend()
plt.show()
```

**Histogram question:**  
Is it generally more fuel efficient to drive in the city or on the highway?


**Histogram answer:** As seen in the histogram above, it is generally more fuel efficient to drive on the highway in comparison to the city.

### 3. Lineplot

```python
# Lineplot (Wheelbase vs turning circle)

# Sort the values to ensure all of the dots are not connected by a line
df_sorted = df.sort_values(by='Wheelbase')

plt.plot(df_sorted['Wheelbase'], df_sorted['Turn.circle'], label="Trend")

# Enter the details of the graph
plt.title("Lineplot of wheelbase bs turning circle")
plt.suptitle("")
plt.xlabel("Wheelbase")
plt.ylabel("Turning Circle")
plt.show()
```

**Lineplot question:**  
What is this relationship? What happens when the wheelbase gets larger?


**Lineplot answer:**  As the wheelbase increases in size, the turning circle increases. Thus, the relationship between the two is a positive linear correlation.


### 4. Barplot

```python
# Barplot for mean horsepower
# group the type of car with their respective horsepower, then determine the mean 
mean_horsepower_by_type = df.groupby('Type')['Horsepower'].mean()
mean_horsepower_by_type.plot(kind='bar')

# Enter the details of the graph
plt.title("Bar Plot of Mean Horsepower by car type")
plt.xlabel("Car Type")
plt.ylabel("Mean Horsepower")
plt.show()
```

**Barplot question:**  
Does a larger car mean more horsepower?



**Barplot answer:** Looking at the barplot above, as the sizer of the car increases the horsepower also increases. Thus, it can be assumed that a larger car would have more horsepower.
