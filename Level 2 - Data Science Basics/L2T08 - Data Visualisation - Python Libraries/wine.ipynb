```python
# Import Libraries
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Load data
df = pd.read_csv('wine.csv')
```

```python

# Filter dataset for the specific wine varieties
filtered_df = df[df['variety'].isin(['Cabernet Sauvignon', 'Pinot Noir', 'Chardonnay'])]

# Create the multi-plot grid with histograms using seaborn
g = sns.FacetGrid(filtered_df, col="variety")
g.map(sns.histplot, "points")

# Add the titles and labels to the multi-plot grid, make use of
# {col_name} to name each histogram respective to its type
g.figure.suptitle("Histogram of Points for Selected Wine Varieties")
g.set_axis_labels("Points", "Count")
g.set_titles("{col_name} Wines")

# Initial issue of subtitles being underneath title:
# Using sources from stack overflow, it was found that
# these titles can be adjusted using subplots_adjust 
plt.subplots_adjust(top=0.85)

# Print the plot
plt.show()

```
