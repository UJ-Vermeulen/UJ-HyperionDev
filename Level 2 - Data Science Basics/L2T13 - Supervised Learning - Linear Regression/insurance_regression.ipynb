# Import the necessary libraries

```python
# Import necessary libraries
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
# Load the required dataset
data = pd.read_csv('insurance.csv')
```

# Scatter Plot of Age vs Charges


```python
# Obtain the relevant columns from the dataset
age = data['age'].values.reshape(-1, 1)
charges = data['charges'].values

# Plot a scatter plot with age on the x-axis and charges on the y-axis
plt.scatter(age, charges, color='blue')
plt.xlabel('Age')
plt.ylabel('Charges')
plt.title('Scatter plot of Age vs Charges')
plt.show()
```

# Scatter Plot of Age vs Charges with best fit line and predictions

```python
# Fit a linear regression model
model = LinearRegression()
model.fit(age, charges)

# Make the predictions
predictions = model.predict(age)

# Plot the scatter plot with the best-fit line
plt.scatter(age, charges, color='blue')
plt.plot(age, predictions, color='red', linewidth=2)
plt.xlabel('Age')
plt.ylabel('Charges')
plt.title('Scatter plot of Age vs Charges with Best-Fit Line')
plt.show()
```

# Conclusion


Age has a positive impact on charges, as seen in the scatter plot above. Meaning as age increases, so does charges.
