```python
# Import pandas
import pandas as pd
```

```python
# Create a DataFrame with balance.txt
df = pd.read_csv('balance.txt',sep=' ')
```

```python
# Compare the average income based on ethnicity.
average_income_ethnicity = df.groupby('Ethnicity')['Income'].mean()
print('\nThe average income based on ethnicity: ')
print(average_income_ethnicity)
```

```python
# On average, do married or single people have a higher balance?
average_balance_married = df.groupby('Married')['Balance'].mean()
print('\nDo married or single people have higher balances?: ')
print(average_balance_married)
```

```python
# What is the highest income in our dataset?
highest_income = df['Income'].max()
print(f'\nThe highest income is: {highest_income}')
```

```python
# What is the lowest income in our dataset?
lowest_income = df['Income'].min()
print(f'\nThe lowest income is: {lowest_income}')
```

```python
# How many cards do we have recorded in our dataset? (Hint: use `sum()`)
total_cards = df['Cards'].sum()
print(f'The total amount of cards is: {total_cards}')
```

```python
# How many females do we have information for vs how many males?
total_gender_information = df['Gender'].value_counts()
print('\nAmount of entries per female vs male: ')
print(total_gender_information)
```
