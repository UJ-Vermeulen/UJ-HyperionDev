**Import Libraries**

```python
# Import libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn import preprocessing
from sklearn.metrics import confusion_matrix, accuracy_score, precision_score, recall_score
```

**Import the Data set**

```python
# Import the dataset
iris = load_iris()
data = pd.DataFrame(iris.data, columns=iris.feature_names)
data['Species'] = iris.target
data.head()
```

**Identify the independent variables x**

```python
# Identify the dependant and independent variables
X = data.drop(columns=['Species'])
y = data['Species'].map({0: 0, 1: 1, 2: 1})
```

**Split the data into a Training and Test set**

```python
# Split the data into training and test set (75% and 25%)
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=0)

print("Training set class distribution:\n", y_train.value_counts())
print("Test set class distribution:\n", y_test.value_counts())
```

**Fit a regression model and predictions**

```python
# Fit the model
regression_model = LogisticRegression()
regression_model.fit(X_train, y_train)

# make predictions on test data
y_pred = regression_model.predict(X_test)
```

**Generate a confusion matrix**

```python
# Create a confusion matrix
conf_matrix = confusion_matrix(y_test, y_pred)
print("Confusion Matrix:\n", conf_matrix)
```

**Confusion Matrix comments:**

As seen in the confusion matrix above, 13 True negatives were found, meaning 13 iris-setosa were correctly predicted, while 0 were incorrectly predicted. There were also 25 True Positives, stating that 25 non-iris-setosa were correctly predicted. This could indicate a model with perfect accuracy as it predicted all instances correctly

**Calculate the accuracy, precision, and recall**

```python
# Calculate accuracy, precision, and recall using {}_score function
# After research the score functions were found to be the simplest
# Way to calculate these metrics with the binary classification results
# of the logistic regression model
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred,)
recall = recall_score(y_test, y_pred,)

print(f"Accuracy: {accuracy}")
print(f"Precision: {precision}")
print(f"Recall: {recall}")
```

**Calculated metrics comments:**

As seen above, the Accuracy, precision, and Recall are all 1 (100%). This substantiates earlier claims of this model having perfect accuracy as all instances were correctly predicted with no mistakes.
